# From Data to Factors

## 1. Introduction

Empirical factor investing starts with a simple idea:

> Securities can be grouped according to characteristics, and the returns on those groups can be used to study systematic patterns in expected returns.

Examples of such characteristics include firm size, valuation ratios, profitability, investment, momentum, volatility, or any other variable that may help explain differences in future returns.

The empirical workflow is therefore not specific to any one factor model. It is a general procedure that takes us from raw security-level data to portfolios, factor returns, and ultimately asset-pricing tests.

A useful way to think about the process is

$$
\text{Data}
\rightarrow
\text{Signal}
\rightarrow
\text{Portfolio Sort}
\rightarrow
\text{Portfolio Return}
\rightarrow
\text{Factor}
\rightarrow
\text{Evaluation}.
$$

This notebook provides a high-level overview of that process.

## 2. Start with the investment universe

The first step is to define which securities are included in the analysis.

In practice, a raw equity database contains many observations that we may not want to treat as ordinary investable common stocks. Researchers therefore typically define an investment universe using filters related to, for example,

- security type,
- exchange,
- listing status,
- availability of return data,
- availability of accounting information,
- and the period over which the analysis is conducted.

The exact definition of the universe matters.

A factor constructed from large, liquid stocks may behave very differently from one constructed from the entire equity market, particularly if very small firms receive substantial weight.

The first principle is therefore simple:

> Before constructing portfolios, define clearly which securities are eligible to enter them.

## 3. Define the characteristic or signal

Once the universe has been defined, each security is assigned a characteristic or signal.

Let

$$
x_{i,t}
$$

denote the value of a characteristic for security \(i\) at time \(t\).

Examples could include

$$
\text{Size}_{i,t},
\qquad
\text{Book-to-Market}_{i,t},
\qquad
\text{Momentum}_{i,t},
\qquad
\text{Profitability}_{i,t}.
$$

The signal is the variable used to rank securities.

For the Fama–French application, the principal quantities are constructed as

**Delisting-adjusted return**

$$
R^{\mathrm{adj}}_{i,t}=(1+R_{i,t})(1+DLRET_{i,t})-1
$$

**Market equity**

$$
ME_{i,t}=\lvert PRC_{i,t}\rvert\times SHROUT_{i,t}
$$

**Book equity**

$$
BE_{i,y}=SEQ_{i,y}+TXDITC_{i,y}-PS_{i,y}
$$

**Book-to-market**

$$
BM_{i,y}=\frac{BE_{i,y-1}}{ME^{\mathrm{Dec}}_{i,y-1}}
$$

**Operating profitability**

$$
OP_{i,y}
=\frac{REVT_{i,y-1}-COGS_{i,y-1}-XSGA_{i,y-1}-XINT_{i,y-1}}
{BE_{i,y-1}}
$$

**Investment**

$$
INV_{i,y}=\frac{AT_{i,y-1}-AT_{i,y-2}}{AT_{i,y-2}}
$$

Here $DLRET$ is the delisting return; when only the ordinary return or delisting return is observed, the available component is retained. $PS$ uses preferred-stock redemption value when available, then liquidation value, then carrying value, and signed CRSP prices enter market equity in absolute value. Nonpositive book equity is excluded from accounting-characteristic sorts. The calendar subscripts emphasize the economic timing: accounting information from fiscal year $y-1$ and December market equity from year $y-1$ are assigned to the June formation in year $y$. June market equity determines the size sort, while lagged monthly market equity supplies the value weights used to earn returns after formation.

Portfolios formed in June of year $y$ are held from July of year $y$ through June of year $y+1$. This convention separates the information date, formation date, and return-realization period.

At this stage, one of the most important empirical issues is **information timing**.

When constructing a portfolio at time \(t\), we must only use information that would actually have been available at time \(t\).

Using information that became available later introduces look-ahead bias.

Conceptually,

$$
\text{Information observed before }t
\rightarrow
\text{Portfolio formed at }t
\rightarrow
\text{Future return measured after }t.
$$

This timing discipline is fundamental to any credible backtest.

## 4. Rank securities and form portfolios

The next step is to rank securities according to the signal.

Suppose we have five securities:

| Security | Signal |
|---|---:|
| A | 0.10 |
| B | 0.25 |
| C | 0.45 |
| D | 0.70 |
| E | 0.90 |

The securities can be ordered from the lowest to the highest signal value.

We may then divide them into groups.

For example,

$$
P_1,\ P_2,\ldots,P_5
$$

could represent five portfolios from low to high values of the characteristic.

With more securities, researchers commonly use

- terciles,
- quintiles,
- deciles,
- or other percentile-based portfolios.

The underlying idea is always the same:

$$
x_{i,t}
\rightarrow
\text{Rank}
\rightarrow
\text{Portfolio assignment}.
$$

This procedure is usually referred to as a **portfolio sort**.

## 5. Breakpoints

To form portfolios, we need rules that determine where one portfolio ends and the next begins.

These thresholds are called **breakpoints**. Let $\mathcal B_t$ denote the reference universe used to calculate them. For requested cumulative probabilities $0<p_1<\cdots<p_{K-1}<1$, define

$$
b_{k,t}=Q_{p_k}\!\left(\left\{x_{i,t}:i\in\mathcal B_t\right\}\right),
\qquad k=1,\ldots,K-1
$$

The first portfolio contains observations at or below the first breakpoint, interior portfolios lie between adjacent breakpoints, and the final portfolio contains observations above the last breakpoint. A median breakpoint therefore divides securities into two groups:

$$
P_{1,t}=\left\{i:x_{i,t}\leq Q_{0.5}(\mathcal B_t)\right\},
\qquad
P_{2,t}=\left\{i:x_{i,t}>Q_{0.5}(\mathcal B_t)\right\}
$$

Similarly, decile portfolios are based on the 10th, 20th, ..., 90th percentiles of the signal distribution.

An important design choice is the population used to calculate these breakpoints.

Breakpoints may be calculated using

- the entire investment universe,
- a particular exchange,
- large and liquid firms,
- or some other reference universe.

This choice can materially affect the resulting portfolios.

## 6. One-dimensional and multidimensional sorts

A portfolio sort does not need to use only one characteristic.

### Univariate sort

A univariate sort ranks securities using a single signal.

For example,

$$
\text{Momentum}
\rightarrow
10\text{ momentum portfolios}.
$$

This is the simplest form of portfolio formation.

### Double sort

Researchers may also want to control for a second characteristic.

Suppose we are interested in profitability but want to ensure that the resulting portfolios are not simply driven by firm size.

We might therefore sort firms according to both

$$
\text{Size}
$$

and

$$
\text{Profitability}.
$$

This creates portfolios at the intersections of the two characteristics.

For example, a \(5\times5\) sort creates

$$
5\times5=25
$$

portfolios.

Multidimensional sorts are widely used because they allow researchers to study one characteristic while controlling for another.

## 7. Independent versus dependent sorts

There are two important ways to perform a multidimensional sort.

### Independent sort

In an independent sort, breakpoints for each characteristic are calculated separately.

For example,

1. determine size breakpoints,
2. determine profitability breakpoints,
3. assign firms according to both sets of breakpoints.

The two classifications are therefore independent.

### Dependent sort

In a dependent or sequential sort, securities are first sorted according to one variable.

Within each resulting group, securities are then sorted according to the second variable.

For example,

$$
\text{Size sort}
\rightarrow
\text{Profitability sort within each size group}.
$$

These procedures need not produce the same portfolios.

The distinction is therefore important whenever more than one characteristic is used.

## 8. Calculate portfolio returns

After assigning securities to portfolios, we calculate the return earned by each portfolio.

The general portfolio return is

$$
R_{p,t}
=
\sum_{i\in p}w_{i,t-1}R_{i,t},
$$

where

- \(R_{i,t}\) is the return on security \(i\),
- \(w_{i,t-1}\) is the weight assigned to the security before the return is realized.

Two common weighting schemes are equal weighting and value weighting.

### Equal-weighted portfolio

Each security receives the same weight:

$$
w_{i,t-1}=\frac{1}{N_{p,t}},\qquad i\in p.
$$

The return is therefore

$$
R_{p,t}
=
\frac{1}{N_{p,t}}
\sum_{i\in p}R_{i,t}.
$$

Equal weighting gives relatively more importance to small firms.

### Value-weighted portfolio

Securities receive weights proportional to their market capitalization:

$$
w_{i,t-1}
=
\frac{ME_{i,t-1}}
{\sum_{j\in p}ME_{j,t-1}}.
$$

Larger firms therefore receive greater weight.

Neither method is universally correct. They answer somewhat different economic questions.

## 9. From portfolios to factors

A factor is often constructed as a long–short combination of portfolio returns.

The simplest example is

$$
F_t
=
R_t^{High}
-
R_t^{Low}.
$$

The factor goes long securities with high values of a characteristic and short securities with low values.

If the characteristic is positively related to expected returns, we would expect

$$
E[F_t]>0.
$$

More generally, let $\mathcal L$ and $\mathcal S$ denote the sets of long- and short-leg portfolios. A normalized long–short factor is

$$
F_t
=
\sum_{p\in\mathcal L}a_pR_{p,t}
-
\sum_{p\in\mathcal S}b_pR_{p,t},\qquad
\sum_{p\in\mathcal L}a_p=\sum_{p\in\mathcal S}b_p=1.
$$

The weights in each leg sum to one, while the signed weights of the combined strategy sum to zero. The factor is therefore a self-financing, dollar-neutral portfolio return. Subtracting the risk-free rate from both legs cancels, so the factor return is already an excess return. Averaging several portfolios on each side can help diversify firm-specific noise and control for other characteristics.

### Worked example: the Fama–French five factors

The market factor is the value-weighted market return in excess of the risk-free rate:

$$
MKT_t=R^{\mathrm{VW}}_{M,t}-R_{f,t}
$$

Following the standard construction, the characteristic factors use value-weighted returns on independently formed Size × B/M, Size × OP, and Size × INV portfolios. With $S$ and $B$ denoting small and big firms, the characteristic spreads are

$$
HML_t=\tfrac{1}{2}(SH_t+BH_t)-\tfrac{1}{2}(SL_t+BL_t),
$$

$$
RMW_t=\tfrac{1}{2}(SR_t+BR_t)-\tfrac{1}{2}(SW_t+BW_t),
$$

$$
CMA_t=\tfrac{1}{2}(SC_t+BC_t)-\tfrac{1}{2}(SA_t+BA_t).
$$

Each characteristic sort also produces a small-minus-big spread:

$$
SMB_t^{BM}=\tfrac{1}{3}(SL_t+SM_t+SH_t)-\tfrac{1}{3}(BL_t+BM_t+BH_t),
$$

$$
SMB_t^{OP}=\tfrac{1}{3}(SW_t+SN_t+SR_t)-\tfrac{1}{3}(BW_t+BN_t+BR_t),
$$

$$
SMB_t^{INV}=\tfrac{1}{3}(SA_t+SN_t+SC_t)-\tfrac{1}{3}(BA_t+BN_t+BC_t).
$$

The published size factor averages these three components:

$$
SMB_t=\tfrac{1}{3}\left(SMB_t^{BM}+SMB_t^{OP}+SMB_t^{INV}\right).
$$

Within a two-letter cell label, the first letter identifies size and the second identifies the characteristic portfolio. Thus $SH_t$ is the value-weighted return in month $t$ on small, high-B/M firms. The remaining labels are $L/M/H$ for low/middle/high B/M, $W/N/R$ for weak/neutral/robust profitability, and $A/N/C$ for aggressive/neutral/conservative investment. The cell portfolios—not the individual stocks—receive equal weights in the factor formulas. Because B/M, OP, and INV impose different data-availability requirements, the three sorts may use different local formation universes even when their timing and eligibility conventions are otherwise identical.

The important point is that factor construction is fundamentally a portfolio-design problem.

Different choices regarding

- the signal,
- breakpoints,
- number of portfolios,
- weighting scheme,
- rebalancing frequency,
- and control variables

can all lead to different factor returns.

## 10. Factors and test portfolios are different objects

A useful distinction is between **factors** and **test portfolios**.

A factor is typically designed to capture a systematic return pattern.

A test portfolio is an asset whose return we use to evaluate an asset-pricing model.

Suppose we construct 25 portfolios using a \(5\times5\) sort.

These portfolios may serve as test assets:

$$
R_{1,t},R_{2,t},\ldots,R_{25,t}.
$$

We can then ask whether a set of factors explains their returns.

This distinction becomes important when moving from portfolio construction to asset-pricing tests.

## 11. From constructed returns to empirical tests

Construction is only the first stage. Average returns, volatility, cumulative performance, drawdowns, and Sharpe ratios describe how a portfolio behaved, but they do not reveal whether an existing model already explains that behavior. Empirical asset pricing therefore complements descriptive evidence with time-series regressions.

The dependent variable must be an excess return. For a long-only portfolio, $R^e_{i,t}=R_{i,t}-R_{f,t}$. A self-financing long–short factor is already an excess return because the risk-free return cancels between its two legs.

### 11.1 Factor-spanning tests

Notebook 2 treats a candidate factor as the asset to be explained and asks whether benchmark factors span its return:

$$
R^e_{G,t}=\alpha+\boldsymbol{\beta}'\mathbf F_t+\varepsilon_t.
$$

The central hypothesis is

$$
H_0:\alpha=0
\qquad\text{against}\qquad
H_1:\alpha\neq0.
$$

Alpha is average return not explained by the benchmark, the loadings measure exposure to its factors, and adjusted $R^2$ measures time-series fit while penalizing additional regressors. Rejecting zero alpha means that the benchmark fails to span the candidate over the sample; it does not by itself distinguish omitted risk from mispricing.

The coefficients are estimated by OLS. Because financial-return residuals may be heteroskedastic and autocorrelated, Notebooks 2 and 3 use Newey–West standard errors with 12 monthly lags. This changes estimated uncertainty and $t$-statistics—not the OLS coefficient estimates. Twelve lags are a transparent one-year convention rather than a universal rule.

### 11.2 Testing a complete asset-pricing model

Notebook 3 reverses the emphasis. Rather than searching for a factor with incremental alpha, it asks whether a proposed model explains a diversified cross-section of $N$ test portfolios. For every portfolio $i$,

$$
R^e_{i,t}=\alpha_i+\boldsymbol{\beta}_i'\mathbf F_t+\varepsilon_{i,t},
\qquad i=1,\ldots,N.
$$

A successful model should leave economically small and statistically insignificant pricing errors. Individual tests use $H_0:\alpha_i=0$, while factor-loading tests show whether a portfolio is exposed to the return dimensions represented by the model. Significant loadings do not, on their own, prove that those factors are structural sources of priced risk.

Model performance is summarized from several complementary perspectives:

$$
A(|\alpha|)=\frac{1}{N}\sum_{i=1}^{N}|\widehat{\alpha}_i|,
\qquad
A(\bar R^2)=\frac{1}{N}\sum_{i=1}^{N}\bar R_i^2.
$$

Average absolute alpha measures the typical pricing-error magnitude, whereas average adjusted $R^2$ measures explanatory power. High fit does not guarantee correct pricing, so neither statistic is sufficient alone.

The Gibbons–Ross–Shanken test evaluates the joint restriction

$$
H_0:\alpha_1=\alpha_2=\cdots=\alpha_N=0
\qquad\text{against}\qquad
H_1:\text{at least one }\alpha_i\neq0.
$$

Rejecting $H_0$ means that the model fails to jointly price that test-portfolio family. Failure to reject is not proof that the model is true. The conventional GRS test also relies on stronger distributional assumptions and is not HAC, so it is interpreted alongside Newey–West individual-alpha tests and the economic magnitude of pricing errors.

## 12. The full empirical workflow

The entire procedure can therefore be summarized as

$$
\boxed{
\text{Define universe}
\rightarrow
\text{Construct signal}
\rightarrow
\text{Form portfolios}
\rightarrow
\text{Calculate returns}
\rightarrow
\text{Construct factor}
\rightarrow
\text{Measure performance}
\rightarrow
\text{Test asset-pricing implications}
}
$$

Each step contains empirical choices.

These choices should be explicit, economically motivated, and implemented using information that would have been available to the investor at the time.

That is the core discipline behind empirical factor investing.
