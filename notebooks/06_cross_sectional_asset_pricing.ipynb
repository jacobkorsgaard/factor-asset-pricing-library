# Cross-Sectional Asset Pricing

Cross-sectional asset pricing asks whether there is a common price for each risk exposure in our asset-pricing models. The weak relationship between market beta and average returns is a well-known empirical challenge to CAPM, illustrated by Fama and French (1992). I use this example to illustrate that problem and explain how cross-sectional pricing tests work. The limitation concerns CAPM as a model, rather than the cross-sectional regression method, which remains in use.

Notebook 05 evaluated candidate factors against a market benchmark. Here I turn to a cross-section of portfolios to examine whether greater market exposure is associated with higher average excess returns, as CAPM predicts. In this notebook, I first explain the pricing restriction and the two-pass estimation procedure. I then apply the method to 25 published Size × Book-to-Market portfolios, keeping market beta as the single exposure so that the estimated relationship can be examined directly.

```python
import sys
from pathlib import Path
ROOT=Path.cwd().resolve()
ROOT=ROOT.parent if ROOT.name=='notebooks' else ROOT
sys.path.insert(0,str(ROOT/'src'))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
NAVY, BLUE, GREEN, ORANGE, GRAY = '#17324D', '#2878A5', '#168A86', '#D99A32', '#737D87'
STYLE={'font.family':'DejaVu Sans','font.size':11,'axes.titlesize':12,'axes.labelsize':11,
       'xtick.labelsize':10,'ytick.labelsize':10,'legend.fontsize':10,'figure.dpi':110,
       'figure.facecolor':'white','axes.facecolor':'white','axes.spines.top':False,
       'axes.spines.right':False,'axes.prop_cycle':plt.cycler(color=[BLUE,GREEN,ORANGE,NAVY]),
       'grid.alpha':.15,'lines.linewidth':1.8}
plt.rcParams.update(STYLE)
SIGNED=LinearSegmentedColormap.from_list('ocean_white_amber',[BLUE,'white',ORANGE])
from factor_asset_pricing.asset_pricing_tests import two_pass_fama_macbeth
from factor_asset_pricing.french_data import load_french, sample_coverage
RAW = ROOT / "data/raw/french"
```

## 1. Cross-sectional asset pricing

### 1.1 From factor exposure to expected return

A time-series regression tells us how a portfolio moves with a factor. An asset-pricing model adds a restriction on its expected return: portfolios with the same exposure should receive the same model-implied compensation for that exposure. In a general linear model,

$$E[R_{i,t}-R_{f,t}]=\boldsymbol\beta_i'\boldsymbol\lambda,$$

where $R_{i,t}$ is portfolio $i$'s total return in month $t$, $R_{f,t}$ is the matching risk-free return, $\boldsymbol\beta_i$ contains portfolio $i$'s factor exposures, and $\boldsymbol\lambda$ contains the prices of risk. A portfolio can move strongly with a factor even when the associated expected compensation is small.

For CAPM, this becomes

$$E[R_{i,t}-R_{f,t}]=\beta_{i,M}\lambda_M,\qquad\lambda_M=E[MKT].$$

Under the model, another unit of market beta earns another unit of the expected market premium. The security market line expresses this relation across investments, with beta on the horizontal axis and expected excess return on the vertical axis. It differs from 05's scatterplots of contemporaneous monthly returns.

### 1.2 Estimating exposures and their compensation

We do not observe the population betas or expected returns. Two-pass estimation lets us estimate them using historical returns.

**First pass: estimate exposures.** For each portfolio, estimate the time-series regression introduced in 05:

$$R_{i,t}-R_{f,t}=\alpha_i+\beta_{i,M}MKT_t+\varepsilon_{i,t}.$$

The estimated $\widehat\beta_{i,M}$ describes its market exposure. We retain the time-series intercept when estimating beta, rather than imposing the pricing restriction before examining it. In this notebook, each first-pass regression is estimated once over the full sample. The estimated betas are then held fixed, without rolling or expanding windows.

**Second pass: compare returns across exposures.** In each month, regress the portfolios' realized excess returns on those same fixed estimated betas:

$$R_{i,t}-R_{f,t}=\gamma_{0,t}+\widehat\beta_{i,M}\lambda_{M,t}+\eta_{i,t}.$$

The slope tells us how much returns differ per unit of beta in that month. It can be positive in one month and negative in another. The common intercept is the fitted excess return at zero beta, not an individual portfolio's time-series alpha. Allowing it makes the fitted relationship more flexible than the strict CAPM restriction, which requires zero intercept and a market exposure price equal to the expected market premium.

### 1.3 Fama–MacBeth estimates and inference

The Fama–MacBeth procedure averages the monthly cross-sectional coefficients (Fama and MacBeth, 1973):

$$\widehat\lambda_M=\frac1T\sum_{t=1}^T\widehat\lambda_{M,t},\qquad
\widehat\gamma_0=\frac1T\sum_{t=1}^T\widehat\gamma_{0,t}.$$

Each monthly slope describes the cross-section observed in that month. Its average estimates persistent compensation per unit of market exposure. For a balanced panel with fixed betas and identical regressors each month, this average equals the slope obtained by regressing the portfolios' sample mean excess returns on those betas. The repeated monthly estimates additionally supply a time series for inference.

If monthly slopes were independent and identically distributed, the standard error of their mean would be

$$SE(\widehat\lambda_M)=\frac{s(\widehat\lambda_{M,t})}{\sqrt T}.$$

Dependence between months changes the uncertainty about that mean. We therefore use the Newey–West approach from 05, applied to the monthly coefficient series with 12 lags. The reported statistic is the coefficient mean divided by its HAC standard error, using an asymptotic reference distribution. There is another source of uncertainty to keep in mind: we estimated the betas in the first pass. This is important to be aware of when performing the cross-sectional regressions.

## 2. Example: pricing the Size × Book-to-Market portfolios

### 2.1 Data and portfolio exposures

For this example, I use the 25 published value-weighted Size × B/M portfolios and market excess returns from the [Kenneth French Data Library](https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/data_library.html). These are published returns, rather than the internal portfolios constructed in Notebook 02. I use these portfolios because they give us a familiar cross-section with differences in size and valuation. Rows run from small to big firms and columns from low to high book-to-market. This portfolio design preserves characteristic-related return differences while reducing firm-specific noise. It does not guarantee a wide range of market betas.

I keep the existing sample from July 1964 through December 2024, giving 726 consecutive monthly observations. The sample requires complete observations for all 25 portfolios and the supplied factor dataset, preserving the previous sample-selection convention even though the regression uses MKT alone. I estimate one market beta for each of the 25 portfolios using the full July 1964–December 2024 history. Keeping those betas fixed, I then run one cross-sectional regression in each of the 726 months. Since the betas use the full history, this is a retrospective pricing exercise rather than a real-time strategy.

The portfolio panel contains total returns. I subtract each month's matched RF once to obtain portfolio excess returns. MKT is already an excess return and is left unchanged. Calculations use decimal returns, while tables and figures express monthly means and exposure prices in percent.

```python
factors = load_french("ff5", RAW).loc["1963-07":"2024-12"]
assert len(factors) == 738 and not factors.isna().any().any()
total = load_french("size_bm", RAW).loc["1964-07":"2024-12"]
sample = pd.concat([total, factors], axis=1, join="inner").dropna()
excess = sample[total.columns].sub(sample["RF"], axis=0)
display(sample_coverage(excess))
print("Empirical monthly common sample:", sample.index.min(), sample.index.max(), "T =", len(sample))
```

```python
fm=two_pass_fama_macbeth(excess,sample[["MKT"]],hac_lags=12,add_constant=True)
betas=fm["betas"]["MKT"]
g=fm["summary"]["estimate"]
observed=excess.mean()
market_mean=sample["MKT"].mean()
fitted=g["constant"]+betas*g["MKT"]
pricing_errors=observed-fitted
asset_order=[f"S{s}C{c}" for s in range(1,6) for c in range(1,6)]
size_colors=[NAVY,BLUE,GREEN,ORANGE,GRAY]
print(f"Estimated market betas: {betas.min():.3f} to {betas.max():.3f}")
```

```python
fig,ax=plt.subplots(figsize=(8.5,4.8),layout="constrained")
grid=np.linspace(betas.min()-.06,betas.max()+.06,150)
ax.plot(grid,100*grid*market_mean,color=NAVY,ls="--",label="CAPM restriction: beta × market mean")
ax.plot(grid,100*(g["constant"]+grid*g["MKT"]),color=GREEN,label="Fitted cross-sectional relationship")
for size,color in enumerate(size_colors,start=1):
    assets=[f"S{size}C{c}" for c in range(1,6)]
    ax.scatter(betas.loc[assets],100*observed.loc[assets],color=color,s=45,label=f"Size {size}",zorder=3)
ax.set(xlabel="Estimated market beta",ylabel="Average excess return (% per month)",title="Market exposure and historical portfolio returns")
ax.legend(fontsize=8,ncol=2,loc="upper center",bbox_to_anchor=(.5,-.16))
plt.show()
```

The plot shows each portfolio's average excess return over the full sample against its fixed estimated beta. It does not show a regression for one particular month. The solid line uses the averages of the monthly cross-sectional intercepts and slopes. With the same portfolios and fixed betas every month, this line is identical to a single regression of full-sample average excess returns on betas. Running the monthly regressions also gives us the coefficient time series needed to estimate standard errors.

The estimated betas range from approximately 0.87 to 1.41, while average excess returns range from approximately 0.29% to 1.10% per month. Portfolios with similar market exposure can therefore have quite different historical means. The dashed line represents the CAPM restriction evaluated using the sample market mean. The solid line instead estimates both an intercept and an exposure price from these portfolios.

What stands out is that the fitted line slopes downward, even though the sample market premium is positive. This is a feature of this selected cross-section, not evidence that greater market exposure generally commands negative expected compensation. The estimated slope's uncertainty is essential to its interpretation, as the table below shows.

### 2.2 Estimated compensation for market exposure

The table brings together the estimated common intercept, the market exposure price, their Newey–West uncertainty, and the sample market premium. For CAPM, the population intercept is zero and the exposure price equals the expected market premium. The market mean gives us a point of comparison. It is important to keep the tests separate: the reported statistics test each coefficient against zero. They do not test whether the estimated exposure price equals the market mean.

```python
summary=fm["summary"].copy()
comparison=pd.DataFrame({
    "Estimate (%)":summary["estimate"]*100,
    "NW SE (%)":summary["standard_error"]*100,
    "NW statistic":summary["t_statistic"],
    "p-value":summary["p_value"],
    "CAPM reference (%)":[0,market_mean*100],
    "Months":summary["n_periods"].astype(int),
})
comparison.index=["Common intercept","Market exposure price"]
display(comparison)
```

The estimated exposure price is approximately −0.308% per month, with a Newey–West standard error of 0.458% and a statistic of −0.67. It is not statistically distinguishable from zero. The sample market premium is positive, approximately 0.576% per month, but the table does not supply a formal test of equality between that mean and the estimated exposure price.

The fitted intercept is approximately 1.073% per month, with a statistic of 2.48. It is significant at the nominal 5% two-sided level. Since none of the portfolios has beta near zero, interpreting it as the return on an observed zero-beta investment would be inappropriate. The intercept and slope together describe a downward fitted relationship over the observed beta range.

These estimates do not reproduce CAPM's sample reference line, and the significant intercept provides evidence against the zero-intercept restriction. The table still does not test whether the estimated exposure price equals the market mean. Nor does insignificance establish that the model is correct. The portfolios' mean-return differences and remaining errors provide additional descriptive evidence.

### 2.3 How well does market exposure explain average returns?

To see how well the fitted relationship explains the portfolios, I plot each portfolio's mean against its freely estimated cross-sectional prediction. Points on the 45-degree line would have zero cross-sectional pricing error. Both axes use the same scale, so vertical deviations show the error in percentage points per month.

```python
fig,ax=plt.subplots(figsize=(6.8,5.4),layout="constrained")
bounds=np.array([min(observed.min(),fitted.min()),max(observed.max(),fitted.max())])*100
pad=.08
limits=[bounds[0]-pad,bounds[1]+pad]
ax.plot(limits,limits,color=GRAY,ls="--",label="Zero pricing error")
for size,color in enumerate(size_colors,start=1):
    assets=[f"S{size}C{c}" for c in range(1,6)]
    ax.scatter(100*fitted.loc[assets],100*observed.loc[assets],color=color,s=45,label=f"Size {size}",zorder=3)
ax.set(xlim=limits,ylim=limits,aspect="equal",xlabel="Fitted mean excess return (% per month)",ylabel="Observed mean excess return (% per month)",title="Pricing errors of the fitted beta relationship")
ax.legend(fontsize=8,loc="lower right")
plt.show()
```

The predictions occupy a narrow range, while observed means vary considerably. The fitted market-beta relationship therefore leaves meaningful differences between these portfolios unexplained. This plot evaluates the estimated cross-sectional line, including its intercept. It is not a plot of time-series CAPM alphas.

The heatmap locates the same errors within the original Size × B/M portfolio design. A positive cell denotes an observed mean above the fitted mean. A negative cell denotes an observed mean below it.

```python
error_grid=pricing_errors.reindex(asset_order).to_numpy().reshape(5,5)*100
lim=np.abs(error_grid).max()
fig,ax=plt.subplots(figsize=(6.8,5),layout="constrained")
im=ax.imshow(error_grid,cmap=SIGNED,norm=TwoSlopeNorm(0,vmin=-lim,vmax=lim),aspect="equal")
for (row,col),value in np.ndenumerate(error_grid):
    ax.text(col,row,f"{value:.2f}",ha="center",va="center",fontsize=10,color="white" if abs(value)>.72*lim else NAVY)
ax.set(xticks=range(5),xticklabels=range(1,6),yticks=range(5),yticklabels=range(1,6),xlabel="Book-to-market quintile (low → high)",ylabel="Size quintile (small → big)",title="Cross-sectional mean-return errors")
fig.colorbar(im,ax=ax,label="Observed − fitted (percentage points per month)")
plt.show()
```

Among the smallest portfolios, the low-B/M portfolio falls approximately 0.346 percentage points per month below the fitted mean, while the high-B/M portfolio sits approximately 0.354 points above it. All five largest-size portfolios have negative errors. The remaining pattern is therefore connected to the portfolios' characteristic layout, rather than merely a single unusual cell.

Average absolute error is approximately 0.143 percentage points per month, and root-mean-squared error is approximately 0.172 points. These describe the magnitude of unexplained historical means. They are neither significance tests nor expected profits available to a future trading strategy. Because the regression includes an intercept, signed errors average to zero across the equally weighted test assets, which is why absolute magnitudes are useful.

### 2.4 What the application establishes

For these portfolios, market beta alone does not explain the pattern in historical mean returns particularly well. The fitted exposure price is imprecise, the fitted means vary little, and structured cross-sectional errors remain. This is a conclusion about these 25 portfolios over this sample, rather than a general claim that market risk never earns compensation. It does not identify whether the departures reflect omitted factors, mispricing, estimation error, or the test-asset design.

I estimate betas from the same full sample used for the cross-sectional comparisons. It is a retrospective pricing exercise, not an implementable sequence of portfolio decisions. The reported HAC uncertainty also conditions on those estimated betas, rather than fully accounting for both estimation stages. The data show what happened in sample, without establishing out-of-sample investment performance.

The two-pass method extends to multiple exposures through the general restriction introduced in Section 1. Here the single market exposure makes the estimation and interpretation visible.
